# Data Analysis

## Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from helpers import load_csv_data
from implementations import mean_squared_error_gd, mean_squared_error_sgd, ridge_regression, logistic_regression, reg_logistic_regression, least_squares
```

## Data Loading


```python
x_train, x_test, y_train, train_ids, test_ids = load_csv_data("dataset")
```

```python
np.random.seed(1)
```

## Data Preprocessing

```python

# feature selection
# todo


# normalisation: keep the training stats, then apply them to both sets
mean = np.nanmean(x_train, axis=0)
std = np.nanstd(x_train, axis=0)
std[std == 0] = 1

x_train = np.nan_to_num((x_train - mean) / std)
x_test = np.nan_to_num((x_test - mean) / std)

x_train = np.hstack((np.ones((x_train.shape[0], 1)), x_train))
x_test = np.hstack((np.ones((x_test.shape[0], 1)), x_test))

# creating a validation set
val_ratio = 0.2
n_samples = x_train.shape[0]
n_val = int(n_samples * val_ratio)
np.random.seed(1)
indices = np.random.permutation(n_samples)
val_idx, train_idx = indices[:n_val], indices[n_val:]
x_val, y_val = x_train[val_idx], y_train[val_idx]
x_train, y_train = x_train[train_idx], y_train[train_idx]



















```

## Hyperparameter Optimisation

```python
# Finding the best hyperparameters for each model:

# - mean squared error
gammas = [0.01, 0.05, 0.1, 0.5, 1, 5, 10]
max_iters = [1000, 5000, 10000]
initial_w = np.zeros(x_train.shape[1])
best_loss = float('inf')
best_gamma_mse_gd = None
best_max_iter_mse_gd = None
for gamma in gammas:
    for max_iter in max_iters:
        w, loss = mean_squared_error_gd(y_train, x_train, initial_w, max_iter, gamma)
        print(f"gamma: {gamma}, max_iter: {max_iter}, loss: {loss}")
        if loss < best_loss:
            best_loss = loss
            best_gamma_mse_gd = gamma
            best_max_iter_mse_gd = max_iter
print(f"mean_squared_error_gd: Best gamma: {best_gamma_mse_gd}, Best max_iter: {best_max_iter_mse_gd}, Best loss: {best_loss}")

# - mean squared error sgd
best_loss = float('inf')
best_gamma_mse_sgd = None
best_max_iter_mse_sgd = None
for gamma in gammas:
    for max_iter in max_iters:
        w, loss = mean_squared_error_sgd(y_train, x_train, initial_w, max_iter, gamma)
        print(f"gamma: {gamma}, max_iter: {max_iter}, loss: {loss}")
        if loss < best_loss:
            best_loss = loss
            best_gamma_mse_sgd = gamma
            best_max_iter_mse_sgd = max_iter
print(f"mean_squared_error_sgd: Best gamma: {best_gamma_mse_sgd}, Best max_iter: {best_max_iter_mse_sgd}, Best loss: {best_loss}")

# - ridge regression
lambdas = [0.01, 0.05, 0.1, 0.5, 1, 5, 10]
best_loss = float('inf')
best_lambda_ridge = None
for lambda_ in lambdas:
    w, loss = ridge_regression(y_train, x_train, lambda_)
    print(f"lambda: {lambda_}, loss: {loss}")
    if loss < best_loss:
        best_loss = loss
        best_lambda_ridge = lambda_
print(f"ridge_regression: Best lambda: {best_lambda_ridge}, Best loss: {best_loss}")

# logistic methods expect labels in {0, 1}
y_train_logistic = (y_train + 1) / 2

# - logistic regression
gammas = [0.01, 0.05, 0.1, 0.5, 1, 5, 10]
max_iters = [1000, 5000, 10000]
best_loss = float('inf')
best_gamma_logistic = None
best_max_iter_logistic = None
for gamma in gammas:
    for max_iter in max_iters:
        w, loss = logistic_regression(y_train_logistic, x_train, initial_w, max_iter, gamma)
        print(f"gamma: {gamma}, max_iter: {max_iter}, loss: {loss}")
        if loss < best_loss:
            best_loss = loss
            best_gamma_logistic = gamma
            best_max_iter_logistic = max_iter
print(f"logistic_regression: Best gamma: {best_gamma_logistic}, Best max_iter: {best_max_iter_logistic}, Best loss: {best_loss}")

# - reg_logistic_regression
initial_w = np.zeros(x_train.shape[1])
max_iters = [1000, 5000, 10000]
gammas = [0.01, 0.05, 0.1, 0.5, 1, 5, 10]
lambdas = [0.01, 0.05, 0.1, 0.5, 1, 5, 10]
best_loss = float('inf')
best_lambda_reg_logistic = None
best_max_iter_reg_logistic = None
best_gamma_reg_logistic = None
for lambda_ in lambdas:
    for max_iter in max_iters:
        for gamma in gammas:
            w, loss = reg_logistic_regression(y_train_logistic, x_train, lambda_, initial_w, max_iter, gamma)
            print(f"lambda: {lambda_}, max_iter: {max_iter}, gamma: {gamma}, loss: {loss}")
            if loss < best_loss:
                best_loss = loss
                best_lambda_reg_logistic = lambda_
                best_max_iter_reg_logistic = max_iter
                best_gamma_reg_logistic = gamma
print(f"reg_logistic_regression: Best lambda: {best_lambda_reg_logistic}, Best max_iter: {best_max_iter_reg_logistic}, Best gamma: {best_gamma_reg_logistic}, Best loss: {best_loss}")   
# - least squares
w, loss = least_squares(y_train, x_train)
print(f"Least squares loss: {loss}")

```
